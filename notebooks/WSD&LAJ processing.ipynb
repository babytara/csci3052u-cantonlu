```python
import hashlib
import os
import urllib.request
from concurrent.futures import ThreadPoolExecutor

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# plot style
plt.rcParams.update({
    "font.family": "DejaVu Sans",
    "font.size": 12,
    "figure.figsize": (8, 6),
    "savefig.dpi": 300,
    "savefig.bbox": "tight",
})
plot_color = ["black", "red", "blue", "yellow"]
alpha = 0.85

show = True
out_dir = "yue-all-nli/data/figures"
cache_dir = "yue-all-nli/data/cache"

# sources
wsd_url = ("https://raw.githubusercontent.com/Aatlantise/canto-nlu/"
           "refs/heads/main/data/wsd/test.jsonl")
wsd_columns = ["sentence1", "sentence2"]
wsd_label = "label"
wsd_target = "target"

laj_comp_url = ("https://raw.githubusercontent.com/Aatlantise/canto-nlu/"
                "refs/heads/main/data/laj/laj_comparison/comparison_test.jsonl")
laj_fine_url = ("https://raw.githubusercontent.com/Aatlantise/canto-nlu/"
                "refs/heads/main/data/laj/laj_dataset.jsonl")
laj_comp_cols = ["sentence1", "sentence2"]
laj_comp_label = "label"
laj_fine_cols = ["sentence"]
laj_fine_label = "label"

sinitic_url = ("https://raw.githubusercontent.com/hannliu/SiniticMTError/"
               "main/cantonese.jsonl")

# io
def load_table(path):
    if path.endswith(".jsonl"):
        return pd.read_json(path, lines=True)
    if path.endswith(".csv"):
        return pd.read_csv(path)
    if path.endswith(".parquet"):
        return pd.read_parquet(path)
    raise ValueError(f"Unsupported file type: {path}")


def cached_load(url):
    os.makedirs(cache_dir, exist_ok=True)
    ext = os.path.splitext(url.split("?")[0])[1] or ".jsonl"
    local = os.path.join(cache_dir, hashlib.md5(url.encode()).hexdigest() + ext)
    if not os.path.exists(local):
        print(f"[download] {url}")
        urllib.request.urlretrieve(url, local)
    else:
        print(f"[cache hit] {local}")
    return load_table(local)


def check_columns(df, cols, name):
    missing = [c for c in cols if c not in df.columns]
    if missing:
        raise KeyError(f"{name}: columns {missing} not found. "
                       f"Available columns: {list(df.columns)}")


# sampling
def sample_rows(df, label_col=None, n=3, seed=0):
    if label_col and label_col in df.columns and df[label_col].nunique() <= n:
        picked = df.groupby(label_col, group_keys=False).apply(
            lambda g: g.sample(1, random_state=seed))
        extra = n - len(picked)
        if extra > 0:
            picked = pd.concat([picked,
                                df.drop(picked.index).sample(extra, random_state=seed)])
        return picked.head(n)
    return df.sample(min(n, len(df)), random_state=seed)


# ------------------------------------------------------------------ plotting
def finish_figure(fig, filename):
    os.makedirs(out_dir, exist_ok=True)
    path = os.path.join(out_dir, filename)
    fig.savefig(path)
    if show:
        plt.show()
    plt.close(fig)
    print(f"[saved {path}]")


def plot_class_balance(df, label_col, name):
    counts = df[label_col].value_counts().sort_index()
    fig, ax = plt.subplots()
    bars = ax.bar(counts.index.astype(str), counts.values,
                  color=plot_color[:len(counts)], alpha=alpha)
    ax.bar_label(bars, labels=[f"{v:,}" for v in counts.values])
    ax.set_xlabel("Label")
    ax.set_ylabel("Number of rows")
    ax.set_title(f"{name} class balance")
    finish_figure(fig, f"{name.lower().replace(' ', '_')}_class_balance.png")

    total = counts.sum()
    big, small = counts.index[0], counts.index[-1]
    return (f"Figure ({name} class balance): label {big} accounts for "
            f"{counts.iloc[0] / total:.1%} of the {total:,} rows "
            f"({counts.iloc[0]:,} versus {counts.iloc[-1]:,} for label {small}), "
            f"a ratio of {counts.iloc[0] / counts.iloc[-1]:.1f}:1.")


def plot_lengths(df, cols, name, bins=30, lengths=None):
    if lengths is None:
        lengths = {c: df[c].astype(str).str.len() for c in cols}
    edges = np.linspace(0, max(s.max() for s in lengths.values()), bins + 1)
    fig, axes = plt.subplots(len(cols), 1, sharex=True, sharey=True)
    axes = np.atleast_1d(axes)
    for i, (ax, (col, s)) in enumerate(zip(axes, lengths.items())):
        ax.hist(s, bins=edges, color=plot_color[i % len(plot_color)], alpha=alpha)
        ax.set_ylabel("Rows")
        ax.set_title(col, fontsize=12)
    axes[-1].set_xlabel("Length (characters)")
    fig.suptitle(f"{name} text length")
    finish_figure(fig, f"{name.lower().replace(' ', '_')}_text_lengths.png")

    parts = [f"{c} has a median of {s.median():.0f} characters (maximum {s.max()})"
             for c, s in lengths.items()]
    return f"Figure ({name} text length, in characters): " + "; ".join(parts) + "."


# ------------------------------------------------------------------ checks
def cleaning_checks(df, text_cols, label_col, name, target_col=None):
    res = {}
    res["rows"] = len(df)
    res["missing_values"] = int(df[text_cols + [label_col]].isna().sum().sum())
    res["empty_strings"] = int(sum((df[c].astype(str).str.strip() == "").sum()
                                   for c in text_cols))
    res["edge_whitespace"] = int(sum((df[c].astype(str) != df[c].astype(str).str.strip()).sum()
                                     for c in text_cols))
    res["duplicate_rows"] = int(df.duplicated(subset=text_cols + [label_col]).sum())
    if len(text_cols) == 2:
        a, b = text_cols
        res["identical_pairs"] = int((df[a] == df[b]).sum())
        key = df[a].astype(str) + "\x01" + df[b].astype(str)
        res["conflicting_labels"] = int(
            df.assign(_k=key).groupby("_k", sort=False)[label_col]
            .nunique().gt(1).sum())
    if "id" in df.columns:
        res["duplicate_ids"] = int(df["id"].duplicated().sum())
    if target_col and target_col in df.columns:
        s1, s2 = text_cols[0], text_cols[1]
        res["target_missing_from_sentence1"] = int(
            sum(t not in s for t, s in zip(df[target_col], df[s1])))
        res["target_missing_from_sentence2"] = int(
            sum(t not in s for t, s in zip(df[target_col], df[s2])))
    res["label_values"] = df[label_col].value_counts().to_dict()

    print("\nCleaning checks")
    for k, v in res.items():
        print(f"   {k}: {v}")
    return res


# ------------------------------------------------------------------ reports
def report(name, df, text_cols, label_col, target_col=None,
           construction=None, task=None):
    line = "=" * 70
    print(f"\n{line}\n{name}  ({len(df):,} rows)\n{line}")
    check_columns(df, text_cols + [label_col], name)

    if construction:
        print("\nConstruction process")
        print("   " + construction.strip())
    if task:
        print(f"\nTask: {task}")

    print("\nSample rows")
    for _, row in sample_rows(df, label_col).iterrows():
        print("   - " + " | ".join(f"{c}: {row[c]}""" for c in df.columns))

    lengths = {c: df[c].astype(str).str.len() for c in text_cols}
    captions = [
        plot_class_balance(df, label_col, name),
        plot_lengths(df, text_cols, name, lengths=lengths),
    ]

    print("\nCaptions (edit before use)")
    for c in captions:
        print("   " + c)

    return cleaning_checks(df, text_cols, label_col, name, target_col)


def report_sinitic_cantonese(name, df):
    line = "=" * 70
    print(f"\n{line}\n{name}  ({len(df):,} rows)\n{line}")
    check_columns(df, ["id", "src", "mt", "ref"], name)

    # flatten first error type / severity
    def first(span_list, key):
        if isinstance(span_list, list) and span_list:
            return span_list[0].get(key)
        return None

    df = df.copy()
    df["n_errors"] = df["annotations"].apply(
        lambda a: len(a.get("annotatedSpans", [])) if isinstance(a, dict) else 0)
    df["first_error_type"] = df["annotations"].apply(
        lambda a: first(a.get("annotatedSpans") if isinstance(a, dict) else None,
                        "error_type"))
    df["first_error_severity"] = df["annotations"].apply(
        lambda a: first(a.get("annotatedSpans") if isinstance(a, dict) else None,
                        "error_severity"))

    print(f"\nRows: {len(df):,}  |  mean errors/row: {df['n_errors'].mean():.2f}  "
          f"|  rows with 0 errors: {(df['n_errors'] == 0).sum()}")

    print("\nSample rows")
    for _, row in df.sample(min(3, len(df)), random_state=0).iterrows():
        print(f"   - id: {row['id']} | src: {row['src']} | mt: {row['mt']} | ref: {row['ref']}")

    counts = df["first_error_type"].fillna("<none>").value_counts()
    fig, ax = plt.subplots()
    bars = ax.bar(counts.index.astype(str), counts.values,
                  color=plot_color[:len(counts)], alpha=alpha)
    ax.bar_label(bars, labels=[f"{v:,}" for v in counts.values])
    ax.set_xlabel("First error type")
    ax.set_ylabel("Number of rows")
    ax.set_title(f"{name} — first-error-type distribution")
    plt.xticks(rotation=30, ha="right")
    finish_figure(fig, f"{name.lower().replace(' ', '_')}_error_types.png")

    caption = (f"Figure ({name} first-error-type distribution): "
               f"{counts.index[0]} is the most frequent first error type "
               f"({counts.iloc[0]:,}/{len(df):,} rows, "
               f"{counts.iloc[0] / len(df):.1%}).")

    lengths = {c: df[c].astype(str).str.len() for c in ["src", "mt", "ref"]}
    caption2 = plot_lengths(df, ["src", "mt", "ref"], name, lengths=lengths)

    print("\nCaptions (edit before use)")
    print("   " + caption)
    print("   " + caption2)

    # light cleaning checks (no label column)
    print("\nCleaning checks")
    print(f"   rows: {len(df)}")
    print(f"   missing src/mt/ref: "
          f"{int(df[['src','mt','ref']].isna().sum().sum())}")
    print(f"   duplicate ids: {int(df['id'].duplicated().sum())}")
    print(f"   error_type values: {df['first_error_type'].value_counts().to_dict()}")
    return df


# ------------------------------------------------------------------ main
if __name__ == "__main__":
    # one parallel fetch on first run; cached afterwards
    with ThreadPoolExecutor(max_workers=3) as ex:
        wsd, laj_comp, laj_fine = ex.map(
            cached_load, [wsd_url, laj_comp_url, laj_fine_url])

    report("WSD", wsd, wsd_columns, wsd_label, target_col=wsd_target)

    print("Completed Report WSD")

    report("LAJ comparison", laj_comp, laj_comp_cols, laj_comp_label)

    print("Completed Report LAJ")

    report("LAJ finetune", laj_fine, laj_fine_cols, laj_fine_label)
    print("Completed Report LAJ finetune")

    sinitic = cached_load(sinitic_url)
    report_sinitic_cantonese("SiniticMTError Cantonese", sinitic)
```
