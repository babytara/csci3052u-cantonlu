```python
import pandas as pd
import json

def load_jsonl(path):
    data = []

    with open(path,'r',encoding='utf=8') as f:
        for line in f:
            if line.strip():
                data.append(json.loads(line))
    
    return pd.DataFrame(data)
```

## Collecting labels from each task specific dataset ##

```python
ld_train = load_jsonl('ld_train.jsonl')
ld_test = load_jsonl('ld_test.jsonl')
ld_val = load_jsonl('ld_val.jsonl')

sent_test = load_jsonl('test.jsonl')
sent_train = load_jsonl('train.jsonl')
sent_val = load_jsonl('valid.jsonl')

nli_sample = load_jsonl('sample_nli.jsonl')

print(ld_train['label'].unique())
print(sent_train['label'].unique())
print(nli_sample['label'].unique())
```
